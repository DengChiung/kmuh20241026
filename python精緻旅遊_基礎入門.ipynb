# 資料結構
Python 有幾種內建的資料結構來存儲和管理資料，包括列表、元組、字典和集合。
* 列表 (List)：
列表是有序且可變的資料結構，使用方括號 [] 來定義。它可以包含不同型別的元素。
* 序對 (Tuple)：
序對和列表相似，但它是不可變的，意思是定義後不能修改。使用小括號 () 來定義。
* 字典 (Dictionary)：
字典是一種鍵值對 (key-value) 的集合，使用大括號 {} 定義，鍵和值之間用冒號 : 隔開。
* 集合 (Set)：
集合是一組無序且不重複的元素，使用大括號 {} 定義。

```python
# 創建字典
fruits = {
    'apple': 7.5,
    'orange': 4.5,
    'banana': 2
}
```

```python

# 1. get(key, default=None): 獲取指定鍵的值
print("get apple:", prices.get('apple'))  # 7.5
print("get grape:", prices.get('grape', 'Not Found'))  # Not Found
```

```python

# 2. keys(): 返回字典中的所有鍵
print("keys:", fruits.keys())  # dict_keys(['apple', 'orange', 'banana'])
```

```python

# 3. values(): 返回字典中的所有值
print("values:", fruits.values())  # dict_values([7.5, 4.5, 2])
```

```python

# 4. items(): 返回字典中的所有 (鍵, 值) 對
print("items:", fruits.items())  # dict_items([('apple', 7.5), ('orange', 4.5), ('banana', 2)])
```

```python

# 5. pop(key, default=None): 移除指定鍵的項目並返回其值
orange_price = fruits.pop('orange')
print("pop orange:", orange_price)  # 4.5
print("after pop:", fruits)  # {'apple': 7.5, 'banana': 2}
```

```python

# 6. popitem(): 移除並返回字典中的最後一對 (鍵, 值)
last_item = fruits.popitem()
print("popitem:", last_item)  # ('banana', 2)
print("after popitem:", fruits)  # {'apple': 7.5}
```

```python

# 7. update(other_dict): 更新字典，添加另一個字典的鍵值對
fruits.update({'grape': 3.0, 'pear': 5.0})
print("update:", fruits)  # {'apple': 7.5, 'grape': 3.0, 'pear': 5.0}
```

```python

# 8. setdefault(key, default=None): 如果鍵不存在，則設置並返回該鍵的值
banana_price = fruits.setdefault('banana', 2.5)
print("setdefault banana:", banana_price)  # 2.5
print("after setdefault:", fruits)  # {'apple': 7.5, 'grape': 3.0, 'pear': 5.0, 'banana': 2.5}
```

```python

# 9. clear(): 移除字典中的所有項目
fruits.clear()
print("clear:", fruits)  # {}
```

```python

# 10. copy(): 返回字典的淺拷貝
original_fruits = {'apple': 7.5, 'orange': 4.5, 'banana': 2}
copied_fruits = original_fruits.copy()
print("copy:", copied_fruits)  # {'apple': 7.5, 'orange': 4.5, 'banana': 2}
```

```python

# 11. fromkeys(iterable, value=None): 創建一個新字典，鍵來自 iterable，值為指定值
new_fruits = dict.fromkeys(['kiwi', 'melon'], 0)
print("fromkeys:", new_fruits)  # {'kiwi': 0, 'melon': 0}
```

```python

# 12. key in dict: 檢查鍵是否在字典中
print("apple in dict:", 'apple' in original_fruits)  # True
```

```python

# 13. del dict[key]: 刪除指定鍵的項目
del original_fruits['orange']
print("after del orange:", original_fruits)  # {'apple': 7.5, 'banana': 2}
```

```python

# 14. len(dict): 返回字典中鍵值對的數量
print("len:", len(original_fruits))  # 2
```

```python
# 創建兩個集合
set1 = {'apple', 'banana', 'orange'}
set2 = {'banana', 'kiwi', 'orange', 'mango'}
```

```python

# 1. add(element): 添加元素到集合中
set1.add('grape')
print("add grape:", set1)  # {'apple', 'banana', 'orange', 'grape'}
```

```python

# 2. remove(element): 移除指定的元素，如果元素不存在，會引發 KeyError
set1.remove('banana')
print("remove banana:", set1)  # {'apple', 'orange', 'grape'}
```

```python

# 3. discard(element): 移除指定的元素，如果元素不存在，不會引發錯誤
set1.discard('orange')
print("discard orange:", set1)  # {'apple', 'grape'}
```

```python

# 4. clear(): 清空集合中的所有元素
set1.clear()
print("clear:", set1)  # set()
```

```python

# 5. union(set2): 返回兩個集合的聯集
union_set = set1.union(set2)
print("union:", union_set)  # {'kiwi', 'orange', 'mango', 'banana'}
```

```python

# 6. intersection(set2): 返回兩個集合的交集
intersection_set = set1.intersection(set2)
print("intersection:", intersection_set)  # {'banana', 'orange'}
```

```python

# 7. difference(set2): 返回集合中不在 set2 中的元素
difference_set = set1.difference(set2)
print("difference:", difference_set)  # {'apple', 'grape'}
```

```python

# 8. issubset(set2): 檢查 set1 是否為 set2 的子集
is_subset = set1.issubset(set2)
print("set1 is subset of set2:", is_subset)  # False
```

```python

# 9. issuperset(set2): 檢查 set1 是否為 set2 的超集
is_superset = set1.issuperset(set2)
print("set1 is superset of set2:", is_superset)  # False
```

```python

# 10. isdisjoint(set2): 檢查兩個集合是否不相交
is_disjoint = set1.isdisjoint(set2)
print("set1 is disjoint with set2:", is_disjoint)  # True
```

```python

# 11. copy(): 返回集合的淺拷貝
set_copy = set2.copy()
print("copy:", set_copy)  # {'banana', 'kiwi', 'orange', 'mango'}
```

```python

# 12. len(set): 返回集合中的元素數量
print("length of set2:", len(set2))  # 4
```

```python

# 13. element in set: 檢查元素是否在集合中
print("apple in set2:", 'apple' in set2)  # False
print("banana in set2:", 'banana' in set2)  # True
```

運算子之優先順序
image.png

### 1. 括號 () (最高優先級)
括號會讓內部的運算先進行，無論其內部運算子的優先順序。

```python
result = (2 + 13) * 4  # 括號內的 2 + 3 先計算，結果是 5 * 4 = 20
print(result)  # Output: 20
```

```python
result = (2 + 13) * 4//5
print(result)
```

### 2. 次方 **
次方運算比乘除法優先。

```python
result = 2 ** 3 ** 2  # 先計算 3 ** 2 (即 9)，再計算 2 ** 9，結果是 512
print(result)  # Output: 512
```

### 3. 正負號 +, -
正負號（單一運算子）比乘除優先。

```python
result = -2 ** 2  # 相當於 -(2 ** 2)，結果是 -4
print(result)  # Output: -4
```

### 4. 乘除法 *, /, //, %
乘法、除法、整數除法、取餘數具有相同優先順序，且它們會從左到右依序計算。

```python
result = 10 / 2 * 3  # 先計算 10 / 2 = 5，再計算 5 * 3 = 15
print(result)  # Output: 15
```

### 5. 加減法 +, -
加法和減法優先順序較低，同樣從左到右計算。



```python
result = 10 - 2 + 3  # 先計算 10 - 2 = 8，再計算 8 + 3 = 11
print(result)  # Output: 11
```

### 6. 比較運算子 <, >, <=, >=, ==, !=
比較運算子比邏輯運算子（and, or）優先。

```python
result = 3 + 4 > 5  # 先計算 3 + 4 = 7，再比較 7 > 5，結果為 True
print(result)  # Output: True
```

### 7. 邏輯運算子 and, or
and 比 or 優先。

```python
result = True or False and False  # 相當於 True or (False and False)，結果為 True
print(result)  # Output: True
```

### 綜合範例
這是一個綜合了多個運算子優先順序的範例：

```python
result = 3 + 5 * 2 ** 2 > 4 and  (5 > 3)  # 括號、次方、乘法、加法、比較、邏輯運算
# 1. 先計算 2 ** 2 = 4
# 2. 再計算 5 * 4 = 20
# 3. 再計算 3 + 20 = 23
# 4. 23 > 4 為 True
# 5. 5 - 3 = 2，not 2 為 False ，因 Python 將非零的整數視為 True，而 0 視為 False
# 6. True and False 結果為 False
print(result)  # Output: False
```

# Python 邏輯封裝
## 什麼是邏輯封裝

在 Python 中，邏輯封裝是一種透過將程式邏輯分成模組化的方式來提升可讀性和重複使用性的方法。常見的邏輯封裝方法有兩種：函式（Function）與物件導向程式設計（OOP, Object-Oriented Programming）。它們讓程式碼更加結構化，並且便於維護和擴展。


## 1. 使用函式進行邏輯封裝

### 什麼是函式？

函式是用來封裝一段可重複使用的程式碼的區塊。你可以將程式邏輯包裝進函式中，並根據需要多次呼叫該函式。函式可以接收輸入參數並返回結果。

### **定義與使用函式的範例**


```python
# 定義一個函式來計算兩個數的和
def add_numbers(a, b):
    return a + b

# 呼叫函式
result = add_numbers(5, 3)
result1 = add_numbers(15, 13)
print(result)  # 輸出: 8
print(result1)
```

### **帶有預設參數的函式**

你可以為函式定義預設參數，當呼叫函式時如果不提供該參數，預設值會被使用。


```python
# 帶有預設參數的函式
def greet(name="Guest"):
    print(f"Hello, {name}!")

# 呼叫時不提供參數，使用預設值
greet()  # 輸出: Hello, Guest!

# 呼叫時提供參數，覆蓋預設值
greet("Alice")  # 輸出: Hello, Alice!
```

### 返回多個值的函式
函式還可以返回多個值，這在需要同時返回多個計算結果時非常有用。


```python
# 定義一個函式返回兩個值
def get_min_max(numbers):
    return min(numbers), max(numbers)

# 呼叫函式並同時接收多個返回值
min_value, max_value = get_min_max([2, 5, 1, 8])
print(min_value, max_value)  # 輸出: 1 8
```

## 區域變數與全域變數
在 Python 中，變數可以根據其作用範圍（Scope）分為區域變數和全域變數。這兩者的區別在於它們的定義位置及存取範圍。了解這兩者對於撰寫模組化、可維護的程式碼非常重要。

### 區域變數 (Local Variables)
區域變數是定義在函數或區塊內的變數，僅在該函數或區塊內可用。當函數執行完畢後，區域變數會被銷毀，不再可用。

**區域變數的特點**

* 定義在函數內部。
* 只能在定義它的函數內使用。
* 當函數執行完畢後，區域變數會被銷毀。


```python
def my_function():
    x = 10  # 區域變數 x
    print(x)  # x 在函數內有效

my_function()  # 會輸出 10

print(x)  # 錯誤：因為 x 是區域變數，無法在函數外部使用
```

### 全域變數 (Global Variables)
全域變數是定義在函數外部的變數，可以在整個程式中存取。全域變數在定義後，可以被任何函數或區塊內的程式碼存取或修改。

**全域變數的特點**
* 定義在函數或程式的外部。
* 可在整個程式中存取，包括函數內。
* 需要使用 global 關鍵字才能在函數內修改全域變數的值。

```python
y = 20  # 全域變數

def another_function():
    print(y)  # 在函數內存取全域變數 y

another_function()  # 會輸出 20
print(y)  # 會輸出 20
```

### 在函數內修改全域變數
如果要在函數內修改全域變數，必須使用 global 關鍵字來告訴 Python 這個變數是全域變數，否則 Python 會將其視為新的區域變數。

**使用 global**



```python
z = 30  # 全域變數

def modify_global():
    global z  # 使用 global 關鍵字
    z = 50  # 修改全域變數 z

modify_global()
print(z)  # 會輸出 50，因為 z 在函數內被修改
```

**不使用 global**


```python
a = 100  # 全域變數

def local_assignment():
    a = 200  # 定義一個區域變數 a
    print(a)  # 會輸出 200 (區域變數)

local_assignment()
print(a)  # 會輸出 100 (全域變數未改變)
```

## 2. 使用物件導向進行邏輯封裝

**什麼是物件導向？**
物件導向程式設計（OOP）是一種將程式邏輯封裝在**類別（class）**中的方法。類別是一種自訂的資料結構，能夠封裝屬性（變數）和方法（函式）。透過 OOP，可以模擬現實世界中的物件，並更好地組織和管理程式邏輯。

**物件導向的基礎概念**
* 類別（Class）：定義物件的藍圖，包含屬性和行為。
* 物件（Object）：由類別實例化而來，是類別的具體表現。
* 屬性（Attributes）：物件的特性，類似於變數。
* 方法（Methods）：物件的行為，類似於函式，但與物件緊密結合。


**定義與使用類別的範例**


```python
# 定義一個簡單的類別
class Person:
    # 初始化方法（constructor），用於設置初始屬性
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 方法：介紹自己
    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

# 創建一個 Person 類的物件
person1 = Person("Alice", 25)

# 呼叫物件的方法
person1.introduce()  # 輸出: My name is Alice and I am 25 years old.
```

**封裝與資料隱藏**

在物件導向中，封裝意味著隱藏物件的內部細節，只暴露外部接口來與物件互動。這樣可以避免外部直接修改物件內部的屬性，確保程式更穩定。


```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # 使用雙下劃線表示私有屬性

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        return self.__balance
# 創建一個 BankAccount 物件
account = BankAccount(100)

# 存錢和取錢
account.deposit(50)
account.withdraw(30)
print(account.get_balance())  # 輸出: 120
```

**繼承與多型**

OOP 的另一個強大功能是繼承（Inheritance），允許一個類別繼承另一個類別的屬性和方法。這樣可以讓程式碼更加模組化和可重複使用。


```python
# 定義父類別
class Animal:
    def speak(self):
        print("The animal makes a sound")

# 定義子類別，繼承自 Animal
class Dog(Animal):
    def speak(self):
        print("The dog barks")

# 創建物件並呼叫方法
dog = Dog()
dog.speak()  # 輸出: The dog barks
```

在這裡，Dog 類別繼承了 Animal 類別，並且覆寫了父類別的 speak() 方法，這就是所謂的多型（Polymorphism）。

**假設我們有一個 DataFrame 具有缺失值：**

```python
import pandas as pd
import numpy as np

# 建立範例 DataFrame
data = {'A': [1, 2, np.nan, 4, 5],
        'B': [np.nan, 2, 3, np.nan, 5],
        'C': [1, 2, 3, 4, np.nan]}
df = pd.DataFrame(data)
print("原始資料:\n", df)
```

### 1. 刪除含有缺失值的資料

```python
# 刪除含有任何缺失值的行
df_dropped_na = df.dropna()
print("刪除含有缺失值的資料:\n", df_dropped_na)
```

### 2. 用常數填補（零、均值等）

```python
# 用 0 填補缺失值
df_filled_zero = df.fillna(0)
print("用 0 填補缺失值:\n", df_filled_zero)

import pandas as pd
import numpy as np

# 建立範例 DataFrame
data = {'A': [1, 2, np.nan, 4, 5],
        'B': [np.nan, 2, 3, np.nan, 5],
        'C': [1, 2, 3, 4, np.nan]}
df = pd.DataFrame(data)
print("原始資料:\n", df)
# 用均值填補缺失值
df_filled_mean = df.fillna(df.mean())
print("用均值填補缺失值:\n", df_filled_mean)
```

### 3. 用鄰近值填補

```python
# 使用前一個值填補 (forward fill)
df_ffill = df.fillna(method='ffill')
print("用前一個值填補:\n", df_ffill)

# 使用後一個值填補 (backward fill)
df_bfill = df.fillna(method='bfill')
print("用後一個值填補:\n", df_bfill)
```

### 4. 用插值法填補

```python
# 線性插值
df_interpolated = df.interpolate()
print("用插值法填補缺失值:\n", df_interpolated)
```

### 5. 使用迴歸填補
迴歸填補會根據其他列的數據，使用迴歸模型來預測缺失值。以下是簡單線性迴歸的範例：

```python
from sklearn.linear_model import LinearRegression

# 刪除 A 列和 (B, C) 列都含有缺失值的行
df_clean = df.dropna(subset=['A', 'B', 'C'])

# 訓練集 (X 是 B 和 C 列，y 是 A 列)
X = df_clean[['B', 'C']]
y = df_clean['A']

# 使用線性迴歸模型來預測缺失的值
model = LinearRegression()
model.fit(X, y)

# 預測 A 列中有缺失值的行
X_missing = df[['B', 'C']][df['A'].isna()].dropna()
df.loc[df['A'].isna() & df['B'].notna() & df['C'].notna(), 'A'] = model.predict(X_missing)

print("使用迴歸填補缺失值:\n", df)
```

### 6. 標記缺失值

```python
# 標記缺失值 (新增一列標記是否缺失)
df['A_missing'] = df['A'].isna().astype(int)
print("標記缺失值:\n", df)
```

### 7. 多重插補（Multiple Imputation）
多重插補是更為進階的填補方法，通常用於填補高維資料中的缺失值。這裡使用 IterativeImputer 進行範例：

```python
from sklearn.experimental import enable_iterative_imputer  # 需要啟用這個功能
from sklearn.impute import IterativeImputer

# 使用多重插補方法
imputer = IterativeImputer()
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
print("多重插補後的資料:\n", df_imputed)
```

# 實作演練
本範例中，使用患者的數據集(此數據集為隨機生成之數據)來探索各種生命體徵與健康結果之間的關聯。目標是：

* 處理遺漏或不一致的數據。
* 視覺化變數之間的關係。

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 載入資料
file_path = 'results.xlsx'  #此數據集為隨機生成之數值可能與真實統計情形不符
data = pd.read_excel(file_path)
```

### 1. 查看空值率
檢查每個欄位中有多少空值

```python
missing_data = data.isnull().sum()
print("每個欄位的空值數：\n", missing_data)
```

### 2. 填補空值
使用平均值填補數值欄位的空值(需先處理符號)

```python
data.fillna({
    'AGE': data['AGE'].mean(),
    'Height': data['Height'].mean(),
    'Weight': data['Weight'].mean(),
    'HR': data['HR'].mean(),
    'NBPs': data['NBPs'].mean(),
    'NBPd': data['NBPd'].mean(),
    'T': data['T'].mean(),
    'RR': data['RR'].mean(),
    'SpO2': data['SpO2'].mean(),
    'Sugar': data['Sugar'].mean(),
    'WBC': data['WBC'].mean()
}, inplace=True)
```

### 3. 處理特殊符號

WBC包含非預期的符號，例如">"、"<"，我們可以用 " " 替換它們

```python
data['WBC'] = data['WBC'].replace({'<': '', '>': ''}, regex=True).astype(float)
```

### 4. 刪除缺失值過多的列
若某一行的缺失值過多（比如超過50%），我們可以選擇刪除該列

```python
threshold = len(data.columns) // 2
data_cleaned = data.dropna(thresh=threshold)
```

### 5. 類別資料轉換
將性別欄位轉換成數字，男性(M)轉成1，女性(F)轉成0

```python

data_cleaned['Sex'] = data_cleaned['Sex'].map({'M': 1, 'F': 0, 'Unknown': -1})

# 查看前幾筆處理後的數據
print(data_cleaned.head())
```

### 6. 檢驗資料關係

性別分布

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Sex', data=data_cleaned)
plt.title('Gender Distribution')
plt.show()
```

### 2. 年齡分佈 (Age Distribution)

此直方圖展示病人年齡的分佈情況，並加上核密度估計 (KDE) 曲線，便於查看年齡的集中趨勢。

```python
plt.figure(figsize=(10, 6))
sns.histplot(x='AGE', data=data_cleaned, bins=20, kde=True)
plt.title('Age Distribution')
plt.show()
```
